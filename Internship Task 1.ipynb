### Data analytics questions

```python
# 1. What is the average score in math (G3)?
average_in_maths = data['G3'].mean()
print(f"The average score in maths is {average_in_maths}")
```

```python
# 2. How many students scored above 15 in their final grade (G3)?
n_student_score_above_15 = (data['G3'] > 15).sum()
print(f"There are {n_student_score_above_15} students who scored more than 15")
```

```python
# 3. Is there a correlation between study time (study time) and the final grade (G3)?
correlation = data['studytime'].corr(data['G3'])
print(f"Yes the correlation between studytime and fianl grade are {correlation}")
```

```python
# 4. Which gender has a higher average final grade (G3)?
gender = data.groupby('sex')
average_by_gender = gender['G3'].mean()
print(f"The highest final grade is by {average_by_gender.idxmax()} are {average_by_gender.max()}")
```

### Data visualization

```python
plt.figure(figsize=(5,4))
plt.hist(data['G3'], bins=10, align='mid')
plt.title("Histogram of final grades", fontsize=20)
plt.xlabel("Students", fontsize=15)
plt.ylabel("Students final grade", fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(5,4))
plt.scatter(data['studytime'],data['G3'], color='orange')
plt.title("Study time Vs Final grade", fontsize=20)
plt.xlabel("Students study time", fontsize=15)
plt.ylabel("Students final grade", fontsize=15)
plt.show()
```

```python
plt.figure(figsize=(5, 4))
plt.bar(average_by_gender.index, average_by_gender.values, color=['pink', 'blue'])
plt.title("Male Vs Female", fontsize=20)
plt.xlabel("Average score by male and female", fontsize=15)
plt.ylabel("Average final grade", fontsize=15)
plt.show()
```
